```python
%reload_ext autoreload
%autoreload 2

import warnings
warnings.filterwarnings('ignore')

from sklearn.utils import check_random_state
import pandas as pd
from logging import getLogger
from pathlib import Path
import os
import sys
sys.path.append(os.pardir)

import conf
# Set seeds to make the experiment reproducible
random_state = conf.random_state
random_ = check_random_state(random_state)
```

```python
logger = getLogger(__name__)
logger.info(f"The current working directory is {Path().cwd()}")
```

```python
variable_list = ["num_train", "eta", "gamma", "num_actions", "s"]
for variable in variable_list:
    save_path = Path(f"../concat_result/{variable}/df")
    save_path.mkdir(exist_ok=True, parents=True)


    log_path1 = Path(f"../result/{variable}")
    df_path1 = log_path1 / "df"
    df_path1.mkdir(exist_ok=True, parents=True)
    all_data1 = pd.read_csv(df_path1 / "all_data_results.csv")


    log_path2 = Path(f"../results/{variable}")
    df_path2 = log_path2 / "df"
    df_path2.mkdir(exist_ok=True, parents=True)
    all_data2 = pd.read_csv(df_path2 / "all_data_results.csv")

    log_path3 = Path(f"../results/{variable}")
    df_path3 = log_path3 / "df"
    df_path3.mkdir(exist_ok=True, parents=True)
    all_data3 = pd.read_csv(df_path3 / "all_data_results.csv")

    
    all_data = pd.concat([all_data1, all_data2, all_data3])
    all_data = all_data.drop_duplicates(subset=["seed", variable, "method"], keep="first")
    
    all_data.to_csv(save_path / "all_data_results.csv", index=False)
```
